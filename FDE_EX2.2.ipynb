```python
# Importing the necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load the CSV file
df = pd.read_csv("D:/VS Code/Foundations of Data Engineering/Assignment2/input_data.csv")
heights = df["Height"]
```

```python
# Calculate the PDF and CDF
unique_vals, counts = np.unique(heights, return_counts=True)
pdf = counts / counts.sum()
cdf = np.cumsum(pdf)
```

```python
# Plot the PDF
plt.subplot(1, 2, 1)
plt.bar(unique_vals, pdf)
plt.title("PDF Height Distribution")
plt.xlabel("Height (cm)")
plt.ylabel("P(X = x)")
plt.grid(True)

# Plot the CDF
plt.subplot(1, 2, 2)
plt.step(unique_vals, cdf)
plt.title("CDF Height Distribution")
plt.xlabel("Height (cm)")
plt.ylabel("F(x) = P(X ≤ x)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generate samples
n_samples = 100000
x1 = np.random.uniform(0, 1, n_samples)
x2 = np.random.uniform(0, 1, n_samples)
z = x1 + x2

# Plot histogram of z
plt.hist(z, bins=100, density=True, alpha=0.6, edgecolor='black', label='Simulated Distribution')

# Theoretical PDF of sum of two Uniform(0,1)
x = np.linspace(0, 2, 1000)
pdf = np.where(x < 1, x, 2 - x)  # piecewise definition
pdf = np.clip(pdf, 0, None)      # ensure no negatives

# Plot theoretical PDF
plt.plot(x, pdf, 'r-', lw=2, label='Theoretical PDF (Triangular)')

# Plot settings
plt.title('Distribution of z = x1 + x2 (Uniform[0,1])')
plt.xlabel('z')
plt.ylabel('Density')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from scipy.stats import norm

# Generate data
np.random.seed(0)
n = 100000
x1 = np.random.normal(0, 1, n)
x2 = np.random.normal(0, 1, n)
z = x1 + x2

# Plot histogram of z
plt.hist(z, bins=100, density=True, alpha=0.6, label='Empirical (x1 + x2)')

# Theoretical PDF: mean = 0, std = sqrt(2)
x = np.linspace(-6, 6, 1000)
pdf_theoretical = norm.pdf(x, loc=0, scale=np.sqrt(2))
plt.plot(x, pdf_theoretical, 'r--', label='Theoretical N(0, √2)')

plt.title("Sum of Two Normal Distributions: N(0,1) + N(0,1)")
plt.xlabel("z = x1 + x2")
plt.ylabel("Density")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Generate samples
n = 100000
x1 = np.random.uniform(0, 1, n)
x2 = np.random.uniform(0, 2, n)
z = x1 + x2

# Plot histogram
plt.hist(z, bins=100, density=True, alpha=0.5, edgecolor='black', label='Empirical (x1 + x2)')

# Theoretical PDF (piecewise trapezoid)
x = np.linspace(0, 3, 1000)
pdf = np.piecewise(
    x,
    [x < 1, (x >= 1) & (x < 2), x >= 2],
    [lambda x: x / 2, 0.5, lambda x: (3 - x) / 2]
)

plt.plot(x, pdf, 'r--', label='Theoretical PDF', linewidth=2)

# Labels and display
plt.title("Sum of x1 ~ U(0,1) and x2 ~ U(0,2)")
plt.xlabel("z = x1 + x2")
plt.ylabel("Density")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
# Parameters
mean = 0
std_devs = [0.5, 1, 2, 3]
x = np.linspace(-10, 10, 1000)

# Plot all curves
plt.figure(figsize=(10, 6))

for std in std_devs:
    pdf = norm.pdf(x, loc=mean, scale=std)
    plt.plot(x, pdf, label=f'σ = {std}')

# Plot settings
plt.title('Normal Distribution Curves with Different Standard Deviations')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()
```

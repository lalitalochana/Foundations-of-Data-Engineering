```python
# Importing the necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
```

```python
# Number of tosses
n = 3
# Probability of head for fair coin
p = 0.5

# Values that X can take (0 to 100)
x = np.arange(0, n + 1)

# Compute PMF and CDF
pmf = binom.pmf(x, n, p)
cdf = binom.cdf(x, n, p)

# PMF Plot
plt.subplot(1, 2, 1)
plt.plot(x, pmf, 'b-', lw=2)
plt.title("PMF of X")
plt.xlabel("Number of Heads")
plt.ylabel("P(X = x)")
plt.grid(True)

# CDF Plot
plt.subplot(1, 2, 2)
plt.step(x, cdf, lw=2)
plt.title("CDF of X")
plt.xlabel("Number of Heads")
plt.ylabel("F(x) = P(X ≤ x)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
from scipy.stats import uniform

# Parameters
a, b = 2, 10
size = 1000

# Generate uniform data
data = np.random.uniform(low=a, high=b, size=size)

# === PDF and Histogram Plot ===
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
# Histogram
plt.hist(data, bins=20, density=True, alpha=0.6, edgecolor='black', label='Sample Histogram')
# Theoretical PDF
x = np.linspace(a - 1, b + 1, 1000)
pdf = uniform.pdf(x, loc=a, scale=b - a)
plt.plot(x, pdf, 'r-', lw=2, label='Theoretical PDF')

plt.title('Uniform Distribution: Histogram & PDF')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
plt.grid(True)

# === CDF Plot ===
plt.subplot(1, 2, 2)
# Empirical CDF
sorted_data = np.sort(data)
empirical_cdf = np.arange(1, size + 1) / size
plt.plot(sorted_data, empirical_cdf, label='Empirical CDF', drawstyle='steps-post')
# Theoretical CDF
cdf = uniform.cdf(x, loc=a, scale=b - a)
plt.plot(x, cdf, 'r--', label='Theoretical CDF')

plt.title('Uniform Distribution: CDF')
plt.xlabel('Value')
plt.ylabel('Cumulative Probability')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
from scipy.stats import norm

# Parameters for normal distribution
mean = 0
std_dev = 1
size = 1000

# Generate normally distributed data
data = np.random.normal(loc=mean, scale=std_dev, size=size)

# === PDF and Histogram Plot ===
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
# Histogram
plt.hist(data, bins=20, density=True, alpha=0.6, edgecolor='black', label='Sample Histogram')
# Theoretical PDF
x = np.linspace(mean - 4*std_dev, mean + 4*std_dev, 1000)
pdf = norm.pdf(x, loc=mean, scale=std_dev)
plt.plot(x, pdf, 'r-', lw=2, label='Theoretical PDF')

plt.title('Normal Distribution: Histogram & PDF')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
plt.grid(True)

# === CDF Plot ===
plt.subplot(1, 2, 2)
# Empirical CDF
sorted_data = np.sort(data)
empirical_cdf = np.arange(1, size + 1) / size
plt.plot(sorted_data, empirical_cdf, label='Empirical CDF', drawstyle='steps-post')
# Theoretical CDF
cdf = norm.cdf(x, loc=mean, scale=std_dev)
plt.plot(x, cdf, 'r--', label='Theoretical CDF')

plt.title('Normal Distribution: CDF')
plt.xlabel('Value')
plt.ylabel('Cumulative Probability')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
from scipy.stats import expon

# Parameters for exponential distribution
scale = 1.0  # scale = 1 / lambda
size = 1000

# Generate exponential data
data = np.random.exponential(scale=scale, size=size)

# === PDF and Histogram Plot ===
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
# Histogram
plt.hist(data, bins=20, density=True, alpha=0.6, edgecolor='black', label='Sample Histogram')
# Theoretical PDF
x = np.linspace(0, np.max(data), 1000)
pdf = expon.pdf(x, scale=scale)
plt.plot(x, pdf, 'r-', lw=2, label='Theoretical PDF')

plt.title('Exponential Distribution: Histogram & PDF')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
plt.grid(True)

# === CDF Plot ===
plt.subplot(1, 2, 2)
# Empirical CDF
sorted_data = np.sort(data)
empirical_cdf = np.arange(1, size + 1) / size
plt.plot(sorted_data, empirical_cdf, label='Empirical CDF', drawstyle='steps-post')
# Theoretical CDF
cdf = expon.cdf(x, scale=scale)
plt.plot(x, cdf, 'r--', label='Theoretical CDF')

plt.title('Exponential Distribution: CDF')
plt.xlabel('Value')
plt.ylabel('Cumulative Probability')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Sigmoid Function
# Define the sigmoid function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Derivative of the sigmoid function
def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1 - s)

# Create an array of input values (from -10 to 10)
x = np.linspace(-10, 10, 21)

# Compute values
y_sigmoid = sigmoid(x)
y_derivative = sigmoid_derivative(x)

#Plot the graph
plt.plot(x, y_sigmoid, label='Sigmoid Function')
plt.plot(x, y_derivative, label="Sigmoid Derivative")

plt.title('Sigmoid Function and Its Derivative')
plt.xlabel('Input')
plt.ylabel('Output')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Tanh Function
# Define tanh function
def tanh(x):
    return np.tanh(x)

# Define derivative of tanh
def tanh_derivative(x):
    return 1 - np.tanh(x)**2

# Create an array of input values (from -10 to 10)
x = np.linspace(-10, 10, 21)

# Compute tanh and its derivative
y_tanh = tanh(x)
y_derivative = tanh_derivative(x)

#Plot the graph
plt.plot(x, y_tanh, label='Tanh Function')
plt.plot(x, y_derivative, label='Tanh Derivative')

plt.title('Tanh Function and Its Derivative')
plt.xlabel('Input')
plt.ylabel('Output')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# ReLU Function
# Define ReLU and its derivative
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1, 0)

# Generate input values
x = np.linspace(-10, 10, 21)
y = relu(x)
y_derivative = relu_derivative(x)

#Plot the graph
plt.plot(x, y, label='ReLU Function')
plt.plot(x, y_derivative, label='ReLU Derivative')
plt.title('ReLU Function and Its Derivative')
plt.xlabel('Input')
plt.ylabel('Output')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Leaky ReLU Range
# Leaky ReLU function and derivative
def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

def leaky_relu_derivative(x, alpha=0.01):
    return np.where(x > 0, 1, alpha)

# Input range
x = np.linspace(-10, 10, 21)
y = leaky_relu(x)
y_derivative = leaky_relu_derivative(x)

#Plot the graph
plt.plot(x, y, label='Leaky ReLU Function')
plt.plot(x, y_derivative, label='Leaky ReLU Derivative')

plt.title('Leaky ReLU Function and Its Derivative')
plt.xlabel('Input')
plt.ylabel('Output')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Define the function
def f(x):
    return np.exp(-x)

# Define the derivative
def f_prime(x):
    return -np.exp(-x)

# Generate input values
x = np.linspace(-5, 5, 100)
y = f(x)
dy = f_prime(x)

# Plotting
plt.plot(x, y, label='f(x) = e^(-x)')
plt.plot(x, dy, label="f'(x) = -e^(-x)")
plt.title("Exponential Decay Function and Derivative")
plt.xlabel("x")
plt.ylabel("Value")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Function
def f(x):
    return np.exp(-x**2)

# Derivative
def f_prime(x):
    return -2 * x * np.exp(-x**2)

# Generate input values
x = np.linspace(-3, 3, 100)
y = f(x)
dy = f_prime(x)

# Plotting
plt.plot(x, y, label='f(x) = e^(-x²)')
plt.plot(x, dy, label="f'(x) = -2x·e^(-x²)", linestyle='--')
plt.title("Standard Gaussian Function and Derivative")
plt.xlabel("x")
plt.ylabel("Value")
plt.grid(True)
plt.legend()
plt.show()
```
